```python
import pandas as pd
```

```python
df = pd.DataFrame(raw_data)
```

```python
df
```

```python
# Rename columns
df = df.rename(columns={
    "User ID": "user_id",
    "User Name": "user_name",
    "Join Date": "join_date",
    "Region": "region",
    "Category": "category",
    "Sales Amount": "sales_amount"
})
```

```python
df
```

```python
df.info()
```

```python
# Convert data types of columns
df["user_id"] = df["user_id"].astype("int64")

df["join_date"] = pd.to_datetime(df["join_date"])

df["sales_amount"] = (
    df["sales_amount"].str.replace("$", "", regex=False).str.replace(",", "").astype(float)
)
```

```python
df.info()
```

```python
# Filter records where sales amount > 5000
filtered_sorted_df = (
    df[df["sales_amount"] > 5000]
    .sort_values(by=["region", "sales_amount"], ascending=[True, False])
    .reset_index(drop=True)
)
```

```python
filtered_sorted_df
```

```python
# Group columns by region and category and calculate sum and mean of sales amount
grouped_summary_df = (
    df.groupby(["region", "category"])
    .agg(total_sales=("sales_amount", "sum"), avg_sales=("sales_amount", "mean"))
    .reset_index()
)
```
