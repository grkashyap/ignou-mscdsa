```python
# Generate missing data

import pandas as pd
import numpy as np

# Step 1: Create the DataFrame with missing values
data = {
    'Customer_ID': [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    'Age': [25, 34, np.nan, 45, 29, np.nan, 52, 38, np.nan, 31],
    'Gender': ['Male', 'Female', 'Female', np.nan, 'Male', 'Female', np.nan, 'Male', 'Female', 'Male'],
    'Income_USD': [55000, 62000, 48000, 71000, np.nan, 53000, 89000, np.nan, 61000, 58000],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Chicago', np.nan, 'Miami', 'New York', 'Houston'],
    'Purchase_Amount': [120.5, 250.0, 85.0, np.nan, 310.0, np.nan, 450.0, 95.0, 210.0, 175.0]
}

df = pd.DataFrame(data)

# Step 2: Identify and count missing values
print("--- Missing Values Count ---")
print(df.isnull().sum())

# Step 3: Clean the dataset
# Fill numeric columns with mean/median
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Income_USD'] = df['Income_USD'].fillna(df['Income_USD'].median())
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(df['Purchase_Amount'].median())

# Fill categorical columns with mode or a placeholder
df['Gender'] = df['Gender'].fillna('Unknown')

# Alternatively, drop rows with missing values for specific critical columns
df = df.dropna(subset=['City'])

print("\n--- Cleaned DataFrame ---")
print(df)
print("\n--- Missing Values Count After Cleaning ---")
print(df.isnull().sum())
```
