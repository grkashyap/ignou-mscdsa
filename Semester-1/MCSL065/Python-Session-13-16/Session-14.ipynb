```python
import pandas as pd
import numpy as np
```

```python
# Generate Data
data = {
    'Customer_ID': [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    'Age': [25, 34, np.nan, 45, 29, np.nan, 52, 38, np.nan, 31],
    'Gender': ['Male', 'Female', 'Female', np.nan, 'Male', 'Female', np.nan, 'Male', 'Female', 'Male'],
    'Income_USD': [55000, 62000, 48000, 71000, np.nan, 53000, 89000, np.nan, 61000, 58000],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Chicago', np.nan, 'Miami', 'New York', 'Houston'],
    'Purchase_Amount': [120.5, 250.0, 85.0, np.nan, 310.0, np.nan, 450.0, 95.0, 210.0, 175.0]
}
```

```python
# Convert the data into a pandas data frame
df = pd.DataFrame(data)
```

```python
df
```

```python
# count missing or null values
df.isnull().sum()
```

```python
# fill missing values

df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Gender'] = df['Gender'].fillna('UnKnown')
df['Income_USD'] = df['Income_USD'].fillna(df['Income_USD'].mean())
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(0)
```

```python
df
```

```python
# Drop the row where City is NaN
df = df.dropna(subset=['City'])
```
